````python
%%writefile heterophily.py
import torch
import dgl

g = dgl.load_graphs("/content/comp.dgl")[0][0]
labels = g.nodes['company'].data['label']
num_nodes = g.num_nodes('company')

def global_heterophily_rate(g, etype, labels):
    src, dst = g.edges(etype=etype)
    diff = (labels[src] != labels[dst]).float()
    return diff.mean().item()

def local_heterophily_rate(g, etype, labels, num_nodes, fraud_label=1):
    src, dst = g.edges(etype=etype)

    # out-degree of each source node under this relation
    deg = torch.zeros(num_nodes, dtype=torch.float)
    deg.scatter_add_(0, src, torch.ones_like(src, dtype=torch.float))

    # heterophilic edge count per source node
    diff_mask = (labels[src] != labels[dst]).float()
    het_count = torch.zeros(num_nodes, dtype=torch.float)
    het_count.scatter_add_(0, src, diff_mask)

    deg_safe = deg.clone()
    deg_safe[deg_safe == 0] = 1.0
    ratio_per_node = het_count / deg_safe

    fraud_mask = labels == fraud_label
    total = ratio_per_node[fraud_mask].sum()

    return (total / num_nodes).item()

relations = g.canonical_etypes
print(f"{'Relation':<20}{'#Edges':<12}{'H_edge (global)':<18}{'H_node (local)':<18}")
for r in relations:
    src, dst = g.edges(etype=r)
    h_edge = global_heterophily_rate(g, r, labels)
    h_node = local_heterophily_rate(g, r, labels, num_nodes)
    print(f"{r[1]:<20}{len(src):<12}{h_edge:<18.4f}{h_node:<18.4f}")
````

````python
%%writefile yelpchi_heterophily.py
import scipy.io
import numpy as np

data = scipy.io.loadmat("/content/YelpChi.mat")
labels = data['label'].flatten()
num_nodes = len(labels)

relations = {
    'R-U-R': data['net_rur'],
    'R-T-R': data['net_rtr'],
    'R-S-R': data['net_rsr'],
}

def global_heterophily_rate(adj, labels):
    src, dst = adj.nonzero()
    diff = (labels[src] != labels[dst])
    return diff.mean()

def local_heterophily_rate(adj, labels, num_nodes, fraud_label=1):
    src, dst = adj.nonzero()
    deg = np.zeros(num_nodes)
    het_count = np.zeros(num_nodes)
    np.add.at(deg, src, 1)
    diff_mask = (labels[src] != labels[dst]).astype(float)
    np.add.at(het_count, src, diff_mask)
    deg_safe = deg.copy()
    deg_safe[deg_safe == 0] = 1
    ratio_per_node = het_count / deg_safe
    fraud_mask = labels == fraud_label
    return ratio_per_node[fraud_mask].sum() / num_nodes

print(f"{'Relation':<12}{'#Edges':<12}{'H_edge (global)':<18}{'H_node (local)':<18}")
combined = None
for name, adj in relations.items():
    n_edges = adj.nnz
    h_edge = global_heterophily_rate(adj, labels)
    h_node = local_heterophily_rate(adj, labels, num_nodes)
    print(f"{name:<12}{n_edges:<12}{h_edge:<18.4f}{h_node:<18.4f}")
    combined = adj if combined is None else combined + adj

# All (homo) — union of all relations
h_edge_all = global_heterophily_rate(combined, labels)
h_node_all = local_heterophily_rate(combined, labels, num_nodes)
print(f"{'All (homo)':<12}{combined.nnz:<12}{h_edge_all:<18.4f}{h_node_all:<18.4f}")
````

````python
%%writefile amazon_heterophily.py
import scipy.io
import numpy as np

data = scipy.io.loadmat("/content/Amazon.mat")
labels = data['label'].flatten()
num_nodes = len(labels)

UNLABELED_CUTOFF = 3305
labeled_mask = np.zeros(num_nodes, dtype=bool)
labeled_mask[UNLABELED_CUTOFF:] = True

relations = {
    'U-P-U': data['net_upu'],
    'U-S-U': data['net_usu'],
    'U-V-U': data['net_uvu'],
}

def filter_to_labeled(adj, labeled_mask):
    src, dst = adj.nonzero()
    keep = labeled_mask[src] & labeled_mask[dst]
    return src[keep], dst[keep]

def global_heterophily_rate(src, dst, labels):
    diff = (labels[src] != labels[dst])
    return diff.mean()

def local_heterophily_rate(src, dst, labels, num_nodes, fraud_label=1):
    deg = np.zeros(num_nodes)
    het_count = np.zeros(num_nodes)
    np.add.at(deg, src, 1)
    diff_mask = (labels[src] != labels[dst]).astype(float)
    np.add.at(het_count, src, diff_mask)
    deg_safe = deg.copy()
    deg_safe[deg_safe == 0] = 1
    ratio_per_node = het_count / deg_safe
    fraud_mask = (labels == fraud_label) & labeled_mask
    n_labeled = labeled_mask.sum()
    return ratio_per_node[fraud_mask].sum() / n_labeled

print(f"{'Relation':<12}{'#Edges':<12}{'H_edge (global)':<18}{'H_node (local)':<18}{'Homophily(g)':<14}{'Homophily(l)':<14}")
combined_src, combined_dst = np.array([], dtype=int), np.array([], dtype=int)
for name, adj in relations.items():
    src, dst = filter_to_labeled(adj, labeled_mask)
    n_edges = len(src)
    h_edge = global_heterophily_rate(src, dst, labels)
    h_node = local_heterophily_rate(src, dst, labels, num_nodes)
    print(f"{name:<12}{n_edges:<12}{h_edge:<18.4f}{h_node:<18.4f}{1-h_edge:<14.4f}{1-h_node:<14.4f}")
    combined_src = np.concatenate([combined_src, src])
    combined_dst = np.concatenate([combined_dst, dst])

h_edge_all = global_heterophily_rate(combined_src, combined_dst, labels)
h_node_all = local_heterophily_rate(combined_src, combined_dst, labels, num_nodes)
print(f"{'All (homo)':<12}{len(combined_src):<12}{h_edge_all:<18.4f}{h_node_all:<18.4f}{1-h_edge_all:<14.4f}{1-h_node_all:<14.4f}")
````

````python
%%writefile elliptic_heterophily.py
import pandas as pd
import numpy as np

# Load the three files
classes = pd.read_csv("/content/elliptic_txs_classes.csv")
edgelist = pd.read_csv("/content/elliptic_txs_edgelist.csv")
features = pd.read_csv("/content/elliptic_txs_features.csv", header=None)

# classes.csv has columns: txId, class ('1'=illicit, '2'=licit, 'unknown')
# Filter out "unknown" class per Venus's instruction
labeled = classes[classes['class'] != 'unknown'].copy()
labeled['label'] = labeled['class'].map({'1': 1, '2': 0})  # 1=illicit(fraud), 2=licit(benign)

print("Labeled nodes:", len(labeled))
print("Illicit (fraud):", (labeled['label'] == 1).sum())
print("Licit (benign):", (labeled['label'] == 0).sum())

# Map txId -> integer node index (only for labeled nodes)
txid_to_idx = {tx: i for i, tx in enumerate(labeled['txId'])}
labels = labeled['label'].values
num_nodes = len(labels)

# Filter edges to only those between two labeled nodes
edgelist.columns = ['txId1', 'txId2']
mask = edgelist['txId1'].isin(txid_to_idx) & edgelist['txId2'].isin(txid_to_idx)
filtered_edges = edgelist[mask]

src = filtered_edges['txId1'].map(txid_to_idx).values
dst = filtered_edges['txId2'].map(txid_to_idx).values

print("Edges among labeled nodes:", len(src))

def global_heterophily_rate(src, dst, labels):
    diff = (labels[src] != labels[dst])
    return diff.mean()

def local_heterophily_rate(src, dst, labels, num_nodes, fraud_label=1):
    deg = np.zeros(num_nodes)
    het_count = np.zeros(num_nodes)
    np.add.at(deg, src, 1)
    diff_mask = (labels[src] != labels[dst]).astype(float)
    np.add.at(het_count, src, diff_mask)
    deg_safe = deg.copy()
    deg_safe[deg_safe == 0] = 1
    ratio_per_node = het_count / deg_safe
    fraud_mask = labels == fraud_label
    return ratio_per_node[fraud_mask].sum() / num_nodes

h_edge = global_heterophily_rate(src, dst, labels)
h_node = local_heterophily_rate(src, dst, labels, num_nodes)

print(f"\n{'Relation':<15}{'#Edges':<12}{'H_edge (global)':<18}{'H_node (local)':<18}{'Homophily(g)':<14}{'Homophily(l)':<14}")
print(f"{'payment':<15}{len(src):<12}{h_edge:<18.4f}{h_node:<18.4f}{1-h_edge:<14.4f}{1-h_node:<14.4f}")
````

````python
%%writefile tfinance_heterophily.py
import torch
import dgl

g = dgl.load_graphs("/content/tfinance")[0][0]
labels_onehot = g.ndata['label']
labels = labels_onehot.argmax(dim=1)  # convert one-hot to single class index

print("Label distribution:", torch.bincount(labels))

num_nodes = g.num_nodes()

def global_heterophily_rate(g, labels):
    src, dst = g.edges()
    diff = (labels[src] != labels[dst]).float()
    return diff.mean().item()

def local_heterophily_rate(g, labels, num_nodes, fraud_label=1):
    src, dst = g.edges()
    deg = torch.zeros(num_nodes, dtype=torch.float)
    deg.scatter_add_(0, src, torch.ones_like(src, dtype=torch.float))
    diff_mask = (labels[src] != labels[dst]).float()
    het_count = torch.zeros(num_nodes, dtype=torch.float)
    het_count.scatter_add_(0, src, diff_mask)
    deg_safe = deg.clone()
    deg_safe[deg_safe == 0] = 1.0
    ratio_per_node = het_count / deg_safe
    fraud_mask = labels == fraud_label
    return (ratio_per_node[fraud_mask].sum() / num_nodes).item()

h_edge = global_heterophily_rate(g, labels)
h_node = local_heterophily_rate(g, labels, num_nodes)

print(f"\n{'Relation':<15}{'#Edges':<14}{'H_edge (global)':<18}{'H_node (local)':<18}{'Homophily(g)':<14}{'Homophily(l)':<14}")
print(f"{'homo':<15}{g.num_edges():<14}{h_edge:<18.4f}{h_node:<18.4f}{1-h_edge:<14.4f}{1-h_node:<14.4f}")
````
